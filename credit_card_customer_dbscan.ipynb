```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# dataset source: https://www.kaggle.com/datasets/aryashah2k/credit-card-customer-data
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "aryashah2k/credit-card-customer-data",
    "Credit Card Customer Data.csv",
    pandas_kwargs={"encoding": "UTF-8"},
)

df.head()
```

```python
df['Sl_No'] = df['Sl_No'].astype(str)
df['Customer Key'] = df['Customer Key'].astype(str)
```

```python
df.info()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Avg_Credit_Limit'], kde=True, bins=30, color='green', edgecolor='black')
plt.title('Distribusi Avg Credit Limit Pelanggan')
plt.xlabel('Avg Credit Limit')
plt.ylabel('Frekuensi (Jumlah Pelanggan)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Avg_Credit_Limit'], color='lightblue')
plt.title('Box Plot Avg Credit Limit Pelanggan')
plt.xlabel('Avg Credit Limit')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
